```python


PATH_CHECKPOINTS = "/Volumes/prd/streaming_management/checkpoints"

def reset_env():
    dbutils.fs.rm(PATH_CHECKPOINTS, True)
    spark.sql("DROP TABLE IF EXISTS prd.l_bronze.customers_cdc")
    spark.sql("DROP TABLE IF EXISTS prd.l_silver.customers")
    spark.sql(
        """
        CREATE TABLE IF NOT EXISTS prd.l_bronze.customers_cdc (
            user_id STRING,
            user_type STRING,
            first_name STRING,
            last_name STRING,
            income DOUBLE,
            balance DOUBLE,
            profession STRING,
            birth_date STRING,
            signup_date STRING,
            operation STRING,
            updated_at TIMESTAMP
        )""")

    spark.sql(
        """
        CREATE TABLE IF NOT EXISTS prd.l_silver.customers (
            user_id STRING,
            user_type STRING,
            first_name STRING,
            last_name STRING,
            income DOUBLE,
            balance DOUBLE,
            profession STRING,
            birth_date STRING,
            signup_date STRING
        )""")
    spark.sql("ALTER TABLE prd.l_silver.customers SET TBLPROPERTIES (delta.enableChangeDataFeed = true)")
reset_env()
```

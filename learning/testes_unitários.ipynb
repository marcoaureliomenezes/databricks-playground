```python
from pyspark.testing.utils import (
    assertDataFrameEqual,
    assertSchemaEqual
)

from pyspark.sql.functions import when, lit, col
def add_new_col(df, new, s_col):
    return (
        df
        .withColumn(new,
            when(col(s_col) == 0, lit('Normal'))
            .otherwise(lit('Unknown')))
    )


def test_add_new_col():
    input_data = [(1, 0), (2, 1), (3, -1), (4, None)]
    output_data = [(1, 0, 'Normal'), (2, 1, 'Unknown'), (3, -1, 'Unknown'), (4, None, 'Unknown')]
    df_input = spark.createDataFrame(input_data, ['id', 's_col'])
    df_expected = spark.createDataFrame(output_data, ['id', 's_col', 'new'])
    df_real = add_new_col(df_input, 'new', 's_col')
    assertDataFrameEqual(df_real, df_expected)
    assertSchemaEqual(df_real.schema, df_expected.schema)

test_add_new_col()
```

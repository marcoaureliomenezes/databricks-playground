```python
class DDLOrders:

    def create_bronze_retail_orders(self):
        table_name = "prd.l_bronze.retail_orders"
        return spark.sql(f"""
            CREATE OR REPLACE TABLE {table_name} (
                order_id BIGINT,
                user_id BIGINT,
                product_id INT,
                user_type STRING,
                device STRING,
                traffic_source STRING,
                created_at TIMESTAMP,
                dat_ref_carga STRING,
                ingestion_time TIMESTAMP,
                file_path STRING
            )
            PARTITIONED BY (dat_ref_carga)
            TBLPROPERTIES (
            delta.autoOptimize.optimizeWrite = true,
            delta.autoOptimize.autoCompact = true
            )
        """)

    def create_silver_retail_orders(self):
        table_name = "prd.l_silver.retail_orders"
        spark.sql(f"""
            CREATE OR REPLACE TABLE {table_name} (
                order_id BIGINT,
                user_id BIGINT,
                product_id INT,
                user_type STRING,
                device STRING,
                traffic_source STRING,
                trans_timestamp TIMESTAMP,
                dat_ref_carga STRING
            )
            PARTITIONED BY (dat_ref_carga)
            TBLPROPERTIES (
                delta.autoOptimize.optimizeWrite = true,
                delta.autoOptimize.autoCompact = true
            )
        """)
        spark.sql(f"ALTER TABLE {table_name} ADD CONSTRAINT valid_order_id CHECK (order_id IS NOT NULL)")
        spark.sql(f"ALTER TABLE {table_name} ADD CONSTRAINT valid_user_id CHECK (user_id IS NOT NULL)")
        spark.sql(f"ALTER TABLE {table_name} ADD CONSTRAINT valid_product_id CHECK (product_id IS NOT NULL)")


DDLOrders().create_bronze_retail_orders()
DDLOrders().create_silver_retail_orders()   
```
